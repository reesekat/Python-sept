```python
#1. Arithmetic operators: +,-,*,/,//,%
a=6
b=10
print(a+b) #addition
print(a-b) #subtraction
print(a*b) #multiplication
print(a/b) #division
print(a%b) #modulus/remainder
print(a//b) #floor division
```

```python
#Assignment operators- =, +=,-=,*=,/=,//=,%=
a=6 # = assignment operator
print(a)
a+=6 
print(a)
#a=a+6 -expansion of the compound assignment operator +=
#while using a compound assignment operator value of the variable should already be defined
```

```python
#Comparison operators - ==,!=,>=,>,<,<=
#comparison operators -output will always be either true or false
a=6
b=10

print(a==b)
print(a!=b)
print(a<b)
print(a>b)
print(a<=b)
print(a>=b)
```

```python
#Logical operators - and,or,not

a=6
b=7
c="hello"
d="hello"

print(a==b and c==d) # F and T -> F
print(a!=b and c==d) #T and T -> T
print(a==b or c==d) #F or T ->T
print(not c==d) #not T -> F
```

```python
#Membership operators
#in,not in

a="hello"

print("e" in a)
print("e" in "hello")
print("k" not in a)
```
